```python
# 1. range() 함수를 사용해 1부터 n까지의 합을 구하는 함수를 만드세요.
# hint! for 문을 이용하여 작성합니다.

## 코드시작 ##

def sum_function(n):
    total = 0
    for i in range(1, n + 1):
        total += i
    return total

## 코드종료 ##

sum_function(5)
```

```python
# 2. 다음은 OO대학교 학생들의 수학 성적입니다. 성적의 평균을 구하세요.

scores = [85, 90, 100, 70, 65, 80, 75]
sum = 0
average = 0

## 코드시작 ##
for score in scores:
    sum += score

average = sum / len(scores)

## 코드종료 ##

print(average)
```

```python
# 3. numpy 라이브러리를 활용하여 크기가 10인 랜덤 벡터를 만들고, 그 평균값을 구하세요.

import numpy as np

## 코드시작 ##
rv = np.random.rand(10)

# 평균값 계산
average = np.mean(rv)

## 코드종료 ##

print(average)
```

```python
# 4. numpy 라이브러리를 활용하여 5X5 배열을 만들고, 첫번째 행과 두번째 행의 위치를 바꾸어보세요.

## 코드시작 ##

arr = np.arange(1, 26).reshape(5, 5)

arr[[0, 1]] = arr[[1, 0]]

## 코드종료 ##
arr
```

```python
# 5. numpy 라이브러리를 활용해 다음 두 배열의 유클리드 거리를 구해보세요.

a = np.array([1,2])
b = np.array([4,5])

## 코드시작 ##

# 유클리드 거리 계산
# A(x1, x2), B(y1, y2)일 때 distance = sqrt((y1-x1)**2 + (y2-x2)**2)
distance = np.sqrt(np.sum((b - a) ** 2))

## 코드종료 ##

print("두 배열 사이의 거리 :", distance)
```
